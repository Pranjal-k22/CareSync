## 🧹 1. Clean Data & Handle Missing Values

```python
import os, sys, json, joblib, time
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
plt.rcParams['figure.dpi'] = 140
sns.set_theme(style='darkgrid')
current_dir = Path.cwd()
PROJECT_ROOT = current_dir.parents[1] if current_dir.name == 'notebooks' else (current_dir.parents[0] if current_dir.name == 'ml' else current_dir)
DATA_PATH = PROJECT_ROOT / 'ml/data/cleaned_hospital_deterioration.csv'
if not DATA_PATH.exists(): DATA_PATH = PROJECT_ROOT / 'datasets/cleaned_hospital_data.csv'
print(f'✅ Project Root: {PROJECT_ROOT}')
df = pd.read_csv(DATA_PATH)
print('Missing Values:', df.isnull().sum().to_dict())
# Handle Fahrenheit to Celsius conversion if needed
if 'temperature' in df.columns:
    df['temperature_c'] = df['temperature'].apply(lambda x: (x - 32) * 5/9 if x > 50 else x)
num_cols = df.select_dtypes(include=[np.number]).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
print('Deduplicated Count:', len(df.drop_duplicates()))
display(df.describe().T[['mean', 'std', 'min', '50%', 'max']])
```
