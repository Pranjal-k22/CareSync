# 12. Flu Data Preprocessing & Quality Audit
**Phase 3 & Phase 6 — Data Quality Pipeline & ColumnTransformer Fitting**
Executes quality checks for missing values, duplicates, and impossible clinical values, then builds scikit-learn ColumnTransformer.

```python
import pandas as pd
from pathlib import Path
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
DATA_PATH = Path('../data/cleaned_disease_dataset.csv')
df = pd.read_csv(DATA_PATH)
print(f'Missing Values: {df.isnull().sum().sum()}')
print(f'Duplicate Rows: {df.duplicated().sum()}')
```

```python
num_cols = ['temperature_c', 'heart_rate', 'spo2_pct', 'cough_severity', 'headache_severity', 'fatigue_score', 'chills', 'body_ache', 'sore_throat', 'age']
cat_cols = ['gender']
preprocessor = ColumnTransformer(transformers=[
    ('num', Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())]), num_cols),
    ('cat', Pipeline([('imputer', SimpleImputer(strategy='most_frequent')), ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))]), cat_cols)
])
print('✅ Preprocessing ColumnTransformer Constructed Successfully!')
```
