## 🔮 1. Real-Time Patient Risk Inference

```python
import os, sys, json, joblib, time
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
plt.rcParams['figure.dpi'] = 140
sns.set_theme(style='darkgrid')
current_dir = Path.cwd()
PROJECT_ROOT = current_dir.parents[1] if current_dir.name == 'notebooks' else (current_dir.parents[0] if current_dir.name == 'ml' else current_dir)
DATA_PATH = PROJECT_ROOT / 'ml/data/cleaned_hospital_deterioration.csv'
if not DATA_PATH.exists(): DATA_PATH = PROJECT_ROOT / 'datasets/cleaned_hospital_data.csv'
print(f'✅ Project Root: {PROJECT_ROOT}')
MODEL_PATH = PROJECT_ROOT / 'ml/artifacts/risk_model_v1.joblib'
model = joblib.load(MODEL_PATH)
sample_patient_vitals = pd.DataFrame([{
    'heart_rate': 108.5,
    'respiratory_rate': 22.0,
    'spo2_pct': 88.5,
    'temperature_c': 38.2,
    'systolic_bp': 110.0,
    'diastolic_bp': 65.0,
    'age': 62.0
}])
prob = model.predict_proba(sample_patient_vitals)[0][1]
risk_score = round(prob * 100, 2)
risk_level = 'HIGH RISK' if prob >= 0.50 else ('MEDIUM RISK' if prob >= 0.25 else 'LOW RISK')
print('=== REAL-TIME INFERENCE RESULT ===')
print(f'Patient Risk Score: {risk_score}%')
print(f'Assigned Risk Level: {risk_level}')
```
