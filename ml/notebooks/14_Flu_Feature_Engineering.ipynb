# 14. Flu Feature Engineering
**Phase 8 — Clinical Feature Engineering & Domain Indicators**
Engineers symptom total score, fever severity index, and cardiorespiratory stress ratio.

```python
import pandas as pd
from pathlib import Path
DATA_PATH = Path('../data/cleaned_disease_dataset.csv')
df = pd.read_csv(DATA_PATH)
df['symptom_total'] = df['cough_severity'] + df['headache_severity'] + df['fatigue_score'] + df['chills'] + df['body_ache'] + df['sore_throat']
df['fever_severity'] = df['temperature_c'] * (1 + df['chills'] * 0.2)
print('Engineered Features Created:')
print(df[['symptom_total', 'fever_severity']].describe())
```
