## 🤖 1. Train Baseline Classifiers

```python
import os, sys, json, joblib, time
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
plt.rcParams['figure.dpi'] = 140
sns.set_theme(style='darkgrid')
current_dir = Path.cwd()
PROJECT_ROOT = current_dir.parents[1] if current_dir.name == 'notebooks' else (current_dir.parents[0] if current_dir.name == 'ml' else current_dir)
DATA_PATH = PROJECT_ROOT / 'ml/data/cleaned_hospital_deterioration.csv'
if not DATA_PATH.exists(): DATA_PATH = PROJECT_ROOT / 'datasets/cleaned_hospital_data.csv'
print(f'✅ Project Root: {PROJECT_ROOT}')
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, roc_auc_score
df = pd.read_csv(DATA_PATH)
target = 'deterioration_next_12h'
features = [c for c in ['heart_rate', 'respiratory_rate', 'spo2_pct', 'temperature_c', 'systolic_bp', 'diastolic_bp', 'age'] if c in df.columns]
sub = df.sample(min(15000, len(df)), random_state=42)
X_train, X_test, y_train, y_test = train_test_split(sub[features], sub[target], test_size=0.2, random_state=42, stratify=sub[target])
rf = RandomForestClassifier(n_estimators=50, random_state=42).fit(X_train, y_train)
acc = accuracy_score(y_test, rf.predict(X_test))
auc = roc_auc_score(y_test, rf.predict_proba(X_test)[:, 1])
print(f'Random Forest Baseline Accuracy: {acc*100:.2f}% | ROC-AUC: {auc*100:.2f}%')
```
