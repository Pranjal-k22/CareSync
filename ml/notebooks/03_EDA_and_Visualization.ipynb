## 📊 1. Exploratory Visualizations

```python
import os, sys, json, joblib, time
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
plt.rcParams['figure.dpi'] = 140
sns.set_theme(style='darkgrid')
current_dir = Path.cwd()
PROJECT_ROOT = current_dir.parents[1] if current_dir.name == 'notebooks' else (current_dir.parents[0] if current_dir.name == 'ml' else current_dir)
DATA_PATH = PROJECT_ROOT / 'ml/data/cleaned_hospital_deterioration.csv'
if not DATA_PATH.exists(): DATA_PATH = PROJECT_ROOT / 'datasets/cleaned_hospital_data.csv'
print(f'✅ Project Root: {PROJECT_ROOT}')
df = pd.read_csv(DATA_PATH)
target = 'deterioration_next_12h'
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
df[target].value_counts().plot(kind='bar', ax=axes[0], color=['#10b981', '#f43f5e'], edgecolor='black')
axes[0].set_title('Target Class Distribution', fontweight='bold')
sns.heatmap(df.select_dtypes(include=[np.number]).corr(), annot=True, fmt='.2f', cmap='coolwarm', ax=axes[1], annot_kws={'size': 8})
axes[1].set_title('Pearson Correlation Matrix', fontweight='bold')
plt.tight_layout()
plt.show()
```
