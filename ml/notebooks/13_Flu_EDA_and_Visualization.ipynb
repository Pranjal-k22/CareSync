# 13. Flu EDA & Clinical Visualizations
**Phase 7 — Exploratory Data Analysis (10 Chart Types)**
Generates target bar/donut charts, KDE distributions, violin plots, box plots, and correlation heatmaps.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
sns.set_theme(style='whitegrid')
DATA_PATH = Path('../data/cleaned_disease_dataset.csv')
df = pd.read_csv(DATA_PATH)
fig, ax = plt.subplots(figsize=(7, 4))
sns.countplot(data=df, x='disease_condition', palette='muted', ax=ax)
ax.set_title('Target Condition Distribution')
plt.show()
```
