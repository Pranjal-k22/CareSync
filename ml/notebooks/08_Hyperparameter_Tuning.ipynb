## 🎛️ 1. Grid Search Optimization

```python
import os, sys, json, joblib, time
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
plt.rcParams['figure.dpi'] = 140
sns.set_theme(style='darkgrid')
current_dir = Path.cwd()
PROJECT_ROOT = current_dir.parents[1] if current_dir.name == 'notebooks' else (current_dir.parents[0] if current_dir.name == 'ml' else current_dir)
DATA_PATH = PROJECT_ROOT / 'ml/data/cleaned_hospital_deterioration.csv'
if not DATA_PATH.exists(): DATA_PATH = PROJECT_ROOT / 'datasets/cleaned_hospital_data.csv'
print(f'✅ Project Root: {PROJECT_ROOT}')
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
df = pd.read_csv(DATA_PATH)
target = 'deterioration_next_12h'
features = [c for c in ['heart_rate', 'respiratory_rate', 'spo2_pct', 'temperature_c', 'systolic_bp', 'diastolic_bp', 'age'] if c in df.columns]
sub = df.sample(min(5000, len(df)), random_state=42)
X_train, X_test, y_train, y_test = train_test_split(sub[features], sub[target], test_size=0.2, random_state=42)
param_grid = {'n_estimators': [20, 50], 'max_depth': [5, 10]}
grid = GridSearchCV(RandomForestClassifier(random_state=42), param_grid, cv=3, scoring='roc_auc')
grid.fit(X_train, y_train)
print('Best Parameters:', grid.best_params_)
print(f'Best Validation ROC-AUC: {grid.best_score_*100:.2f}%')
```
