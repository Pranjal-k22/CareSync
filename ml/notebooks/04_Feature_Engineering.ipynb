### *Constructing Shock Index, Pulse Pressure, and Hypoxia-Tachycardia ratios*

## 🛠️ 1. Construct Derived Clinical Features

```python
import os, sys, json, joblib, time
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
plt.rcParams['figure.dpi'] = 140
sns.set_theme(style='darkgrid')
current_dir = Path.cwd()
PROJECT_ROOT = current_dir.parents[1] if current_dir.name == 'notebooks' else (current_dir.parents[0] if current_dir.name == 'ml' else current_dir)
DATA_PATH = PROJECT_ROOT / 'ml/data/cleaned_hospital_deterioration.csv'
if not DATA_PATH.exists(): DATA_PATH = PROJECT_ROOT / 'datasets/cleaned_hospital_data.csv'
print(f'✅ Project Root: {PROJECT_ROOT}')
df = pd.read_csv(DATA_PATH)
spo2_col = 'spo2_pct' if 'spo2_pct' in df.columns else 'spo2'
df['shock_index'] = df['heart_rate'] / df['systolic_bp'].replace(0, np.nan)
df['pulse_pressure'] = df['systolic_bp'] - df['diastolic_bp']
df['hypoxia_tachycardia_ratio'] = df['heart_rate'] / df[spo2_col].replace(0, np.nan)
display(df[['shock_index', 'pulse_pressure', 'hypoxia_tachycardia_ratio']].head(5))
```
